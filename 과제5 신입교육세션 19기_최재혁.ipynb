# 전처리

```python
data_train = data_train.copy()
data_test = data_test.copy()

contract_mapping = {'Monthly': 1, 'Quarterly': 3, 'Annual': 12}
data_train['Contract Length'] = data_train['Contract Length'].map(contract_mapping)
data_test['Contract Length'] = data_test['Contract Length'].map(contract_mapping)


data_train = pd.get_dummies(data_train, drop_first=True)
data_test = pd.get_dummies(data_test, drop_first=True)

train_features = data_train.columns.drop('Churn')
data_test = data_test.reindex(columns=train_features, fill_value=0)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
X = data_train.drop(['Churn', 'CustomerID'], axis=1, errors='ignore')
y = data_train['Churn']
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,         
    max_depth=5,              
    min_samples_split=20,     
    min_samples_leaf=10,      
    random_state=42           
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_val)
```

```python
print(f"검증 데이터 정확도 (Accuracy): {accuracy_score(y_val, y_pred):.4f}")
print("\n[ 분류 성능 상세 리포트 ]")
print(classification_report(y_val, y_pred))
```

```python
X_test = data_test.drop('CustomerID', axis=1, errors='ignore')
```

```python
final_predictions = rf_model.predict(X_test)
```

```python
submission = pd.DataFrame({
    'CustomerID': data_test['CustomerID'],
    'Churn': final_predictions
})
```

```python
print("\n[ 최종 Test 데이터 예측 분포 ]")
print(submission['Churn'].value_counts(normalize=True))
```

```python
conditions_test = (
    (data_test['Age'] >= 51) | 
    (data_test['Payment Delay'] >= 21) | 
    (data_test['Contract Length'] == 1) |  # 아까 1로 매핑했으므로 1을 사용
    (data_test['Total Spend'] < 500)
)

# 조건에 해당하는 사람의 비율 확인
print(f"Test 데이터 내 4가지 조건 해당자 비율: {conditions_test.mean():.4f}")
```

# 100% 비율로 이탈인 데이터들 제거

```python
outlier_conditions = (
    (data_train['Age'] >= 51) | 
    (data_train['Payment Delay'] >= 21) | 
    (data_train['Contract Length'] == 1) | 
    (data_train['Total Spend'] < 500)
)

data_train_clean = data_train[~outlier_conditions].copy()

print(f"제거 전 원본 Train 데이터 수: {len(data_train)}명")
print(f"이상치 제거 후 남은 데이터 수: {len(data_train_clean)}명")

X_clean = data_train_clean.drop(['Churn', 'CustomerID'], axis=1, errors='ignore')
y_clean = data_train_clean['Churn']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

X_train_c, X_val_c, y_train_c, y_val_c = train_test_split(
    X_clean, y_clean, test_size=0.2, stratify=y_clean, random_state=42
)

rf_model_clean = RandomForestClassifier(
    n_estimators=200,         
    max_depth=5,              
    min_samples_split=20,     
    min_samples_leaf=10,      
    random_state=42           
)
rf_model_clean.fit(X_train_c, y_train_c)

y_pred_c = rf_model_clean.predict(X_val_c)

print(f"정제된 데이터 검증 정확도: {accuracy_score(y_val_c, y_pred_c):.4f}")
print("\n[ 분류 성능 상세 리포트 ]")
print(classification_report(y_val_c, y_pred_c))

X_test = data_test.drop('CustomerID', axis=1, errors='ignore')
final_clean_predictions = rf_model_clean.predict(X_test)

submission_clean = pd.DataFrame({'Churn': final_clean_predictions})
print("\n[ 최종 Test 데이터 예측 분포 (이상치 제거 모델) ]")
print(submission_clean['Churn'].value_counts(normalize=True))
```
